# Problem 7

```python
import pandas as pd
import numpy as np
from scipy.stats import norm

df = pd.read_csv("data/ames_prices.csv")

air = (df["Central.Air"] == "Y").astype(int).to_numpy()

p_hat = air.mean()
print(f"n = {len(air)}")
print(f"{p_hat:.4f}")
```

```python

rng = np.random.default_rng(seed=1)   
R = 2000
n = len(air)

boots = np.empty(R)
for i in range(R):
    sample = rng.choice(air, size=n, replace=True)
    boots[i] = sample.mean()

se_hat = boots.std(ddof=1)
print(f"Bootstrap SE: {se_hat:.6f}")
```

```python
z_vals = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
for conf, z in z_vals.items():
    lo = p_hat - z * se_hat
    hi = p_hat + z * se_hat
    print(f"{int(conf*100)}% CI: ({lo:.4f}, {hi:.4f})")
```

```python
theta0 = 0.90
z = (p_hat - theta0) / se_hat
p_value = 2 * norm.cdf(-abs(z))
print(f"z = {z:.3f}")
print(f"two-sided p-value = {p_value:.4f}")

for alpha in [0.10, 0.05, 0.01]:
    reject = p_value < alpha
    print(f"alpha={alpha}: {'Reject' if reject else 'Fail to reject'} H0")
```

```python
theta0 = 0.90
z = (p_hat - theta0) / se_hat
p_value = 2 * norm.cdf(-abs(z))
print(f"z = {z:.3f}")
print(f"two-sided p-value = {p_value:.4f}")

for alpha in [0.10, 0.05, 0.01]:
    reject = p_value < alpha
    print(f"alpha={alpha}: {'Reject' if reject else 'Fail to reject'} H0")
```

Agreement check
* 0.9 is outside of all three intervals, which matches the decision to reject at every alpha.

Interpretation
* 93.31 of the homes in the Ames sample have central air. The boostrap SE is 0.46%, showing that the estimate is precise. All the CI's are above 0.9 and rejects the null. (seed=1, R=2000)

# Problem 8


```python
df8 = df[["area", "price"]].dropna().copy()
x = np.arcsinh(df8["area"].to_numpy())
y = np.arcsinh(df8["price"].to_numpy())

rho_hat = np.corrcoef(x, y)[0, 1]
print(f"n = {len(x)}")
print(f"Estimated correlation: {rho_hat:.4f}")
```

```python
rng = np.random.default_rng(seed=1)
R = 2000
n = len(x)
boots = np.empty(R)

for i in range(R):
    idx = rng.choice(n, size=n, replace=True)  
    boots[i] = np.corrcoef(x[idx], y[idx])[0, 1]

se_hat = boots.std(ddof=1)
print(f"Bootstrap SE: {se_hat:.6f}")
```

```python
z_vals = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
for conf, z in z_vals.items():
    lo = rho_hat - z * se_hat
    hi = rho_hat + z * se_hat
    print(f"{int(conf*100)}% CI: ({lo:.4f}, {hi:.4f})")
```

```python
theta0 = 0
z = (rho_hat - theta0) / se_hat
p_value = 2 * norm.cdf(-abs(z))
print(f"z = {z:.3f}")
print(f"two-sided p-value = {p_value:.4f}")

for alpha in [0.10, 0.05, 0.01]:
    reject = p_value < alpha
    print(f"alpha={alpha}: {'Reject' if reject else 'Fail to reject'} H0")
```

Agreement check
* 0 is outside every interval, agrees with rejecting at every alpha

Interpretation
*   The correlation between area and price is 0.72. All CIs are above 0 and the test reject p=0 at every level. Larger homes tend to have higher prices in this sample.

Why not resample separately?
* Correlation needs the original pairing. Resampling area and price independently breaks the pairs, so the boostrap no longer estimates the real relationship and usually pushes the correlation toward 0. (seed=1, R=2000)

# Problem 9

```python
df9 = pd.read_csv("data/metabric.csv")
time = df9["Overall Survival (Months)"].dropna().to_numpy()

mu_hat = time.mean()
print(f"n = {len(time)}")
print(f"Estimated mean (months): {mu_hat:.2f}")
print(f"Estimated mean (years): {mu_hat/12:.2f}")
```

```python
rng = np.random.default_rng(seed=1)
R = 2000
n = len(time)
boots = np.empty(R)
for i in range(R):
    boots[i] = rng.choice(time, size=n, replace=True).mean()

se_hat = boots.std(ddof=1)
print(f"Bootstrap SE (months): {se_hat:.3f}")
```

```python
z_vals = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
for conf, z in z_vals.items():
    lo = mu_hat - z * se_hat
    hi = mu_hat + z * se_hat
    print(f"{int(conf*100)}% CI (months): ({lo:.2f}, {hi:.2f})")
    print(f"{int(conf*100)}% CI (years):  ({lo/12:.2f}, {hi/12:.2f})")
```

```python
theta0 = 120
z = (mu_hat - theta0) / se_hat
p_value = 2 * norm.cdf(-abs(z))
print(f"z = {z:.3f}")
print(f"two-sided p-value = {p_value:.4f}")

for alpha in [0.10, 0.05, 0.01]:
    reject = p_value < alpha
    print(f"alpha={alpha}: {'Reject' if reject else 'Fail to reject'} H0")
```

Agreement
*  120 is outside all three intervals

Interpretation
* The mean recorded time to death or last follow-up is 129 months. All CIs are above 120 months, so we reject the null of 120 months. Living patients are right-censored, so this is not an estimate of unrestricted mean lifetime. (seed=1, R=2000)

# Problem 10

```python
df10 = pd.read_csv("data/metabric.csv")


df10["deceased"] = (df10["Overall Survival Status"] == "1:DECEASED").astype(int)


yes = df10.loc[df10["Chemotherapy"] == "YES", "deceased"].dropna().to_numpy()
no  = df10.loc[df10["Chemotherapy"] == "NO",  "deceased"].dropna().to_numpy()

delta_hat = yes.mean() - no.mean()
print(f"n_YES = {len(yes)}, n_NO = {len(no)}")
print(f"Estimated difference (YES - NO): {delta_hat:.4f}")
```

```python
rng = np.random.default_rng(seed=1)
R = 2000
boots = np.empty(R)
for i in range(R):
    boot_yes = rng.choice(yes, size=len(yes), replace=True)
    boot_no  = rng.choice(no,  size=len(no),  replace=True)
    boots[i] = boot_yes.mean() - boot_no.mean()

se_hat = boots.std(ddof=1)
print(f"Bootstrap SE: {se_hat:.4f}")
```

```python
z_vals = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
for conf, z in z_vals.items():
    lo = delta_hat - z * se_hat
    hi = delta_hat + z * se_hat
    print(f"{int(conf*100)}% CI: ({lo:.4f}, {hi:.4f})")
```

```python
theta0 = 0
z = (delta_hat - theta0) / se_hat
p_value = 2 * norm.cdf(-abs(z))
print(f"z = {z:.3f}")
print(f"two-sided p-value = {p_value:.4f}")

for alpha in [0.10, 0.05, 0.01]:
    reject = p_value < alpha
    print(f"alpha={alpha}: {'Reject' if reject else 'Fail to reject'} H0")
```

Agreement
* 0 is outside the 90% interval and inside the 95% and 99% intervals

Interpretation
* Patient who received chemoterapy have a slightly lower observed deceased proportion. The result is narrow, significant at 90% but not at 95% or 99% (seed=1, R=2000)

potential confounders are age at diagnosis and tumor stage.

```python
z_vals = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
for conf, z in z_vals.items():
    lo = mu_hat - z * se_hat
    hi = mu_hat + z * se_hat
    print(f"{int(conf*100)}% CI: ({lo:.3f}, {hi:.3f})")
```

```python
theta0 = 8
z = (mu_hat - theta0) / se_hat
p_value = 2 * norm.cdf(-abs(z))
print(f"z = {z:.3f}")
print(f"two-sided p-value = {p_value:.4f}")

for alpha in [0.10, 0.05, 0.01]:
    reject = p_value < alpha
    print(f"alpha={alpha}: {'Reject' if reject else 'Fail to reject'} H0")
```

Agreement 
* 8 is outside every interval

Interpretation
* Mean weekday sleep in this sample is 7.62 hours. All CIs are below the 8 hours. 

The difference is modest (0.4). The result is statistically significant because the sample is large, but the gap is small. (seed=1, R=2000)
